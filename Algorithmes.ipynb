```python
def moyenne_n(n):
    """Version ITERATIVE. Moyenne des entiers 1..n.
    Traduction de l'algoruthme MoyenneN():Saisir(n) devient un paramètre.
    Retourne None si n=0 (cas d'erreur).
    """
    if n==0:
        print("Erreur:n doit être superieur  0")
        return None
    somme = 0
    for i in range (1,n+1): # Pour i=1, n faire
        somme = somme + i
    moyenne = somme/n
    return moyenne
n = 28
print("La moyenne des ",n,"premiers nombres est :")
moyenne_n(n)
    
```

```python
def somme_n_rec(n):
    """Version RECURSIVE de la somme 1+2+...+n."""
    if n == 0:                       # cas de base 
        return 0
    return somme_n_rec(n-1) + n     # cas recursif
    

def moyenne_n_rec(n):
    """Moyenne des entiers 1..n, calculer à partir de la somme récursive."""
    if n == 0:
        print("Erreur : n doit être supérieur à 0")
        return None
    return somme_n_rec(n) / n

n = 3
print("La moyenne des ",n,"premiers nombres est :")
moyenne_n_rec(n)

    
```

```python
def factn(n):
    """Version ITERATIVE de n!"""
    if n == 0 :
       return 1
    fact = 1
    for i in range (1,n + 1) :    # Pour i = 1, n faire
        fact = fact * i
        print("fact=", fact)
    return fact

n = 6
print("Le factoriel de ",n," est (iter) : ", factn(n))
        
```

```python
def fact_rec(n):
    """Version RECURSIVE de n!"""
    if n == 0 :                  # cas de base
       return 1
    return fact_rec(n-1) * n     # cas récursif

n = 4
print("Le factoriel de ",n," est (rec) : ", fact_rec(n))
    
```

```python
def fib_rec(n):
    """Version RECURSUVE (algorythme Fib_Rec)."""
    if n == 0 :
         return 0
    if n == 1 :
         return 1
    return fib_rec(n-1)+ fib_rec(n-2)

n = 4
print("Fibonacci recursif de ",n," est : ", fib_rec(n))

def fib_iter(n) :
    """Version ITERATIVE, en 0(n)."""
    if n == 0 :
        return 0
    a, b = 0, 1
    for _ in range (2, n + 1) :
        a,b = b, a + b
    return b
n = 10
print("Fibonacci itératif de ",n," est : ", fib_iter(n))
```

```python
"""Somme de 2 entiers en base 2"""
def plus_base2(x, y):
    z = x + y
    reste = z - ((x + y) // 2) * 2
    return reste

print(" Somme de 2 entiers en base 2 est : " , plus_base2(3, 7))  
```

```python
""" Somme de 2 entiers en base 2 avec test de parité"""
def plus_base2_v2(x, y):
    z = x + y
    if z % 2 == 0:  # si z est pair
        reste = 0
    else:           # sinon
        reste = 1
    return reste

print(" Somme de 2 entiers en base 2 avec test de parité est :", plus_base2_v2(4, 7))

```

```python
"""Multiplication de 2 entiers en base 2"""
def fois_base2(x,y):
    z = x * y
    reste = z - (z // 2) * 2
    return reste

print(" Multiplication de 2 entiers en base 2 est :", fois_base2(4,7))
```

```python
"""Multiplication de 2 entiers en base 2 avec test de parité"""
def fois_base2_v2(x, y):
    z = x * y
    if z % 2 == 0:  # si z est pair
        reste = 0
    else:           # sinon
        reste = 1
    return reste
    
print(" Multiplication de 2 entiers en base 2 avec test de parité est :", fois_base2_v2(3,5))
```
